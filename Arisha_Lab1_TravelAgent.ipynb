# Travel Agent Example — Arisha (23122106)
Simple rule-based agent demonstrating if / elif / else decision logic with group discounts.

```python
num_passengers = 4
base_price = 15000
destination = "Murree"

if num_passengers >= 5:
    discount = 0.20
elif num_passengers >= 3:
    discount = 0.10
else:
    discount = 0

total_price = base_price * num_passengers * (1 - discount)
print(f"Total price for {num_passengers} passengers to {destination}: PKR {total_price}")
```
